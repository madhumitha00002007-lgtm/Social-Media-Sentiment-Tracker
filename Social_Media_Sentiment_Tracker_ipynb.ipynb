```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Social Media User Feedback Dataset
social_data = {
    "User": ["Abhi", "Balaji", "Charu", "Divya", "Eshwar", "Farooq", "Gokul", "Hari"],
    "Feedback": [
        "I absolutely love this new update! Amazing experience.",
        "Worst customer service ever. Highly frustrated.",
        "It's okay, nothing special but works fine.",
        "Super fast delivery and great quality! Highly recommended.",
        "The app keeps crashing every 5 minutes. Worst app.",
        "Average performance, could be improved a lot.",
        "Wow! This product saved my time. Excellent job!",
        "Not satisfied with the performance. Disappointed.",
    ],
    "Sentiment": [
        "Positive",
        "Negative",
        "Neutral",
        "Positive",
        "Negative",
        "Neutral",
        "Positive",
        "Negative",
    ],
}
df_sentiment = pd.DataFrame(social_data)

print(
    "=================== PROJECT 4: SOCIAL MEDIA SENTIMENT ENGINE ==================="
)
print(df_sentiment[["User", "Sentiment"]])
print("==============================================================================\n")

# 2. Calculating Sentiment Share
metrics = df_sentiment["Sentiment"].value_counts()

# 3. Plotting Pie Chart
plt.figure(figsize=(6, 6))
theme_colors = ["#2ecc71", "#e74c3c", "#f1c40f"]  # Green, Red, Yellow
plt.pie(
    metrics,
    labels=metrics.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=theme_colors,
    explode=(0.05, 0.05, 0),
)
plt.title("Public Sentiment Metrics Share", fontsize=12, fontweight="bold")
plt.show()
```
